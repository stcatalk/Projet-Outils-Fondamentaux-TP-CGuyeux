## Exercice 1.1 : la transposée

Écrivez `transposee(A)` qui échange lignes et colonnes.

**Résultat attendu :** la transposée de `[[1, 2, 3], [4, 5, 6]]` est
`[[1, 4], [2, 5], [3, 6]]`.


```python
# votre réponse ici
def transposee(A):
    nb_lignes = len(A[0])
    nb_colonnes = len(A)
    B = []
    for j in range(nb_lignes):
        B.append([])
        for i in range(nb_colonnes):
            B[j].append(0)
            B[j][i] = A[i][j]
    return B

assert transposee([[1, 2, 3], [4, 5, 6]]) == [[1, 4], [2, 5], [3, 6]]
```

## Exercice 2.1 : produit à la main

Écrivez `produit(A, B)`, et vérifiez sur un exemple que vous savez calculer de tête.

**Résultat attendu :** multiplier une matrice par l'identité la laisse inchangée.


```python
# votre réponse ici
def produit(A, B):
    colonnesA = len(A[0])
    lignesA = len(A)
    colonnesB = len(B[0])
    lignesB = len(B)

    if colonnesA != lignesB:
        raise ValueError

    C = [[0 for _ in range(colonnesB)] for _ in range(lignesA)]

    for ligneC in range(lignesA):
        for colonneC in range(colonnesB):
            C[ligneC][colonneC] = A[colonneC][ligneC] * B[ligneC][colonneC]
    return C
            

A = [
    [0, 0, 1],
    [0, 1, 0],
    [1, 0, 0]
    ]

B = [
    [95, 52, 0],
    [10, 32, 15],
    [8.2, 23, 1]
    ]

print(produit(A,B))
    
```
